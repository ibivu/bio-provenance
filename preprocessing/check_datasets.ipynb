```python
zero_counts = (data == 0).sum() / len(data) * 100

# Bar plot
plt.figure(figsize=(10, 6))
zero_counts.plot(kind='bar')
plt.xlabel('Columns')
plt.ylabel('% Zero values')
plt.title('Zero Value Counts per Column')
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```
